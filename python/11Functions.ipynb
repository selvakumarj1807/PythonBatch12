# Python Function
-----------------

Function is a block of Code. Which only runs. When it is Called

# 01 No Return Type WithOut Argument

```python
# Create the Function

def welcome():
    print('Welcome to Python Functions!...')
```

```python
# Call the Function

welcome()

welcome()

welcome()
```

# 02 No Return Type With Arguments

```python
# Create the Function

def simpleCalculater(a, b):
    print(a + b)
    print(a - b)
    print('')
    
```

```python
# Call the Function

simpleCalculater(10, 2)

simpleCalculater(20, 2)

simpleCalculater(30, 2)
```

# 03 Return Type WithOut Arguments

```python
# Create the Function

def addTwoNum():
    a = int(input('Enter Value 01 : '))

    b = int(input('Enter Value 02 : '))

    c = a + b

    return c
```

```python
# Call the Function

print(addTwoNum())

print('----------------------------------------------')

program01 = addTwoNum()

print(program01)
```

# 04 Return Type With Arguments

```python
# Create the Function

def addTwoNum(a, b):
    c = a + b
    return c
```

```python
# Call the Function

print(addTwoNum(10, 2))

print('----------------------------------------------')

program01 = addTwoNum(20, 2)

print(program01)
```

# Arbitrary Argument Function

```python
# Create the Function

def userDetails(*details):
    print(details)
```

```python
# Call the Function

userDetails(101, 'Sara', 27, 'Female')
```

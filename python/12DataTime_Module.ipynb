# Date Time Module:
-------------------

```python
import datetime as dt
```

# System  Date

```python
currentDate = dt.date.today()

print(currentDate)
```

# System Date & Time

```python
currentDateTime = dt.datetime.now()

print(currentDateTime)
```

# User Date

```python
userDate = dt.date(2025, 10, 25)

print(userDate)
```

# User Date & Time

```python
userDateTime = dt.datetime(2025, 10, 25, 11, 50, 20)

print(userDateTime)
```

# Split Date & Time

```python
currentDateTime = dt.datetime.now()

print(currentDateTime)

print('--------------------------------------')

splitDate = currentDateTime.date()

splitTime = currentDateTime.time()

print('Today : ', splitDate)

print('Time : ', splitTime)
```

# Difference Between Two Dates

```python
currentDate = dt.date.today()

print(currentDate)

print('---------------------------------')

userDate = dt.date(1999, 7,18)

print(userDate)

print('---------------------------------')

diff = currentDate - userDate

print(diff)
```

# Date Time Format

```python
currentDateTime = dt.datetime.now()

print(currentDateTime)
```

```python
format01 = currentDateTime.strftime("%A / %B / %d / %Y")

print(format01)
```

```python
format02 = currentDateTime.strftime("%d / %B / %Y - %A")

print(format02)
```

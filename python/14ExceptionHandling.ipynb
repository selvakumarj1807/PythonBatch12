# Exception Handling or Error Handling

# Try Except

```python
try:
    a = 10
    b = 0
    
    c = a / b
    
    print(c)

except:
    print('Dinominator Cannot be Zero')
```

# Except Exception as e

```python
try:
    a = 10
    b = 0
    
    c = a / b
    
    print(c)

except Exception as e:
    print('Dinominator Cannot be Zero')
    print('The Program have Some Error : ', e)
```

# Try Except Else

```python
try:
    a = 10
    b = 2
    
    c = a / b

except Exception as e:
    print('Dinominator Cannot be Zero')
    print('The Program have Some Error : ', e)

else:
    print(c)
    print('Program run Successfully!..')
```

# Try Except Else Finally

```python
try:
    a = 10
    b = 0
    
    c = a / b

except Exception as e:
    print('Dinominator Cannot be Zero')
    print('The Program have Some Error : ', e)

else:
    print(c)
    print('Program run Successfully!..')

finally:
    print('Thank You. Have a great Day!...')
```

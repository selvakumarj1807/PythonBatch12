# File Handling
---------------

# Read File

```python
readFile = open('test.txt', 'r')

print(readFile.read())

readFile.close()
```

# Write File

```python
writeFile = open('test.txt', 'w')

writeFile.write('This Is File Write Method')

writeFile.close()
```

# Append

```python
appendFile = open('test.txt', 'a')

appendFile.write(' This Is File Append Method')

appendFile.close()
```

# File Read Method

```python
# read

readFile = open('para.txt', 'r')

print(readFile.read())

readFile.close()
```

```python
# readline

readFile = open('para.txt', 'r')

print(readFile.readline())

readFile.close()
```

```python
# readlines

readParaFile = open('para.txt', 'r')

para = readParaFile.readlines()

print(para)

print('-----------------------------------------------------')

print(para[1])

readParaFile.close()
```

# File Delete Method

```python
import os

fileName = 'sample.txt'

if os.path.exists(fileName):
    print('Yes File is There!..')
    os.remove(fileName)
    print(fileName, ' File Deleted Successfully!..')

else:
    print(fileName, ' File is not in ypur Folder!..')
```

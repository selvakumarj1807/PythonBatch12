# Inheritance:
--------------

# 1. Single Inheritance

```python
class College:
    website = 'www.SRS.in'
    collegeName = 'SRS'

class Student(College):
    studentName = 'Sara'
    rollNo = '20PCA101'

studObject01 = Student()
```

```python
print(studObject01.studentName)

print(studObject01.rollNo)

print('-------------------------------------------')

print(studObject01.collegeName)

print(studObject01.website)
```

# Multiple Inheritance

```python
class College:
    website = 'www.SRS.in'
    collegeName = 'SRS'

class Department:
    departmentName = 'English'
    departmentId = 10

class Student(College, Department):
    studentName = 'Sara'
    rollNo = '20PCA101'

studObject01 = Student()
```

```python
print(studObject01.studentName)

print(studObject01.rollNo)

print('-------------------------------------------')

print(studObject01.collegeName)

print(studObject01.website)

print('-------------------------------------------')

print(studObject01.departmentName)

print(studObject01.departmentId)
```

# Multilevel Inheritance

```python
class Father:
    def playing(self):
        print('Father have Playing Chess')

    def job(self):
        print('Doctor')

class Mother(Father):
    def playing(self):
        print('Mother have Playing Chess')

    def cooking(self):
        print('Mother have a cooking Skill')

class Son(Mother):
    def habbit(self):
        print('Son have a Reading Habbit')

sonObject01 = Son()
```

```python
sonObject01.habbit()
```

```python
sonObject01.playing()

sonObject01.cooking()
```

```python
sonObject01.playing()

sonObject01.job()
```

# Abstraction & Encapsulation

```python
# Create the Class

class Library:
    def __init__(self, books):
        self.books = books

    def displayBooks(self):
        for i in self.books:
            print(i, end=', ')

    def borrowBook(self, book):
        if book in self.books:
            print(book, ' Is in your Library!..')
            print('Get your Book Now!...')

            self.books.remove(book)
            
        else:
            print(book, ' Is not in your Librarry!..')

    def reciveBook(self, book):
        self.books.append(book)
        print(book, ' Book Recived Successfully!..')
```

```python
## print('Welcome Our Library Management Project')

books = ['C', 'C++', 'Java']

libraryObject01 = Library(books)

msg = """
1. Display Books
2. Borrow Book
3. Receive Book"""


while True:
    print(msg)

    option = int(input('Select One Option(1, 2, 3): '))
    print('')

    if option == 1:
        libraryObject01.displayBooks()
        print('')

    elif option == 2:
        book = input('Enter your borrowBook Name : ')
        print('')
        libraryObject01.borrowBook(book)
        print('')

    elif option == 3:
        book = input('Enter your Recive Book Name : ')
        print('')
        libraryObject01.reciveBook(book)
        print('')

    else:
        print('Thank you Come again!..')
        break
        
```

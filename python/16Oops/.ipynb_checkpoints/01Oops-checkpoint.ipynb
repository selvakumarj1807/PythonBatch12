# Class

# Declare Variables in Class

```python
class Student:
    name = 'Sara'
    age = 27
```

# Get Values from Class

```python
# getattrribute

print('Student Name : ', getattr(Student, 'name'))

# Get Values using DOT (.) Notation

print('Student Age : ', Student.age)
```

# Difference between getattr & .

```python
# getattrribute

print('Student Gender : ', getattr(Student, 'gender', 'male'))

# cannot do get non declare variable & values
```

# Change Value 

```python
print('Before Student Age : ', Student.age)

Student.age = 50 

print('Aftere Student Age : ', Student.age)
```

# Set or Create Variable after Class Creation

```python
class Staff:
    pass
```

```python
setattr(Staff, 'staffName', 'Sri')

print(Staff.staffName)

Staff.age = 28

print(Staff.age)
```

# Delete Class Variable & Values

```python
class Student:
    name = 'Sara'
    age = 27
```

```python
# Using Del Attribute

print(Student.name)

delattr(Student, 'name')

print(Student.name)
```

```python
print(Student.age)

del(Student.age)

print(Student.age)
```

# Create the Function Inside the Class

```python
class Student:
    def getData(name, age):
        print(name, ' your age is ', age)
        
```

# Call the Function from the Class

```python
# Using Getattr

getattr(Student, 'getData')('Sara', 23)
```

```python
# Using . Notation

Student.getData('Lara', 25)
```

# Class Creation
----------------

```python
class Course:
    courseName = 'Python'
```

# Call the Variable from the Class

```python
print(Course.courseName)
```

# Call the Variable Using the Object

```python
courseObjesct01 = Course()

print(courseObjesct01.courseName)

courseObjesct01.courseName = 'Python Fullstack'

print(courseObjesct01.courseName)
```

```python
courseObjesct02 = Course()

print(courseObjesct02.courseName)

courseObjesct02.courseName = 'Java Fullstack'

print(courseObjesct02.courseName)
```

```python
#using Class

print(Course.courseName)

# Using Object

print(courseObjesct01.courseName)
print(courseObjesct02.courseName)
```

# Create the Function
---------------------

```python
class userData():
    name = 'Sara'
    age = 25
    
    def showUser(self):
        print(self.name, ' your age is ', self.age)
        
```

# Create Object

```python
userObject01 = userData()

userObject01.showUser()
```

```python
print(userObject01.name)

print(userObject01.age)

print('--------------------------------------')

userObject01.name = 'Lara'
userObject01.age = 30

print('--------------------------------------')

print(userObject01.name)

print(userObject01.age)
```

```python
userObject01.showUser()
```

# INIT Method
-------------

```python
class StudentDetails:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def showdetails(self):
        print(self.name, ' your age is ', self.age)
```

```python
studObject01 = StudentDetails('Sara', 27)
```

```python
studObject01.showdetails()
```

```python
studObject02 = StudentDetails('Lara', 22)

studObject02.showdetails()
```

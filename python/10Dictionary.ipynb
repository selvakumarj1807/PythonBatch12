# Dictionary
------------

1. Dictionary is a collection of data type
2. Declare - {"Key": "Value"}
3. It can store multiple values and diffrent datatype values in a single variable. Could not Based on index
4. Dictionary is a mutable
5. Dictionary Remove Duplicate Values based on Key

```python
user = {"name": "Sara", "age": 27, "mobile": "9500912258", "name": "Lara", "name": "Thara"}

print(user, type(user))
```

# Get a Single Value
--------------------

```python
user = {"name": "Sara", "age": 27, "mobile": "9500912258"}

print(user)

print('------------------------------------------------')

print('User Name : ', user['name'])
```

# Change Value
--------------

```python
user = {"name": "Sara", "age": 27, "mobile": "9500912258"}

print(user)

print('------------------------------------------------')

print('User Name : ', user['name'])

print('------------------------------------------------')

user['name'] = 'Lara'

print('User Name : ', user['name'])
```

# add a Single Value or group of Values
---------------------------------------

```python
user = {"name": "Sara", "age": 27, "mobile": "9500912258"}

print(user)

print('------------------------------------------------')

userBank = {"accountHolder": "Sara BOB", "accountNumber": "BOB10290"}

print(userBank)

print('------------------------------------------------')

user.update({"gender": "male"})

print(user)

print('------------------------------------------------')

user.update(userBank)

print(user)
```

# Delete Value
--------------

```python
user = {"name": "Sara", "age": 27, "mobile": "9500912258"}

print(user)

print('------------------------------------------------')

user.pop('mobile')

print('------------------------------------------------')

print(user)
```

```python
user = {"name": "Sara", "age": 27, "mobile": "9500912258"}

print(user)

print('------------------------------------------------')

for i in user:
    print(i, ' - ', user[i])
```

# Fetch Group of values
-----------------------

```python
users = {
    "user01": {"name": "Sara", "age": 27, "mobile": "9500912258"},
    "user02": {"name": "Lara", "age": 22, "mobile": "9080707090"},
    "user03": {"name": "Thara", "age": 27, "mobile": "8754632258"},
    "user04": {"name": "Sam", "age": 12, "mobile": "7080906060"}
}

print(users)
```

```python
print(users['user01'])
```

```python
print(users['user01']['name'])
print(users['user02']['name'])
print(users['user03']['name'])
print(users['user04']['name'])
```

```python
users = {
    "user01": {"name": "Sara", "age": 27, "mobile": "9500912258"},
    "user02": {"name": "Lara", "age": 10, "mobile": "9080707090"},
    "user03": {"name": "Thara", "age": 27, "mobile": "8754632258"},
    "user04": {"name": "Sam", "age": 12, "mobile": "7080906060"}
}

for i in users:    
    print(i, 'Name : ', users[i]['name'])
    print(i, 'Age : ', users[i]['age'])
    print(i, 'Mobile : ', users[i]['mobile'])

    if users[i]['age'] >= 18:
        print('Eligible for Vote')
    else:
        print('No Eligible For Vote')
        
    print('')
```
